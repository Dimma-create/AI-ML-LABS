```python
from sklearn.preprocessing import StandardScaler

print("Before scaling:")
print("Minimum:", X.min(axis=0))
print("Maximum:", X.max(axis=0))
print("Mean:", X.mean(axis=0))
print("Standard deviation:", X.std(axis=0))

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\nAfter scaling:")
print("Minimum:", X_scaled.min(axis=0))
print("Maximum:", X_scaled.max(axis=0))
print("Mean:", X_scaled.mean(axis=0))
print("Standard deviation:", X_scaled.std(axis=0))
```

- Feature scaling is important because SVMs are sensitive to the scale of the input features. Features with larger numerical ranges can have a stronger influence on the geometry of the model and the position of the decision boundary. Standardization puts the features on a comparable scale, allowing the SVM to treat them more appropriately when determining the decision boundary and margin.
- After standardization, each feature is centered around a mean of approximately 0 and scaled to have a standard deviation of approximately 1.

- There are 26 support vectors in the trained Linear SVM model.

- Support vectors are important because they determine the position of the decision boundary and the margin. They are the observations closest to the boundary, so changes to these observations can significantly affect the trained SVM model.

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

C_values = [0.01, 0.1, 1, 10, 100]

for C_value in C_values:
    model = SVC(kernel="linear", C=C_value)
    model.fit(X_scaled, y)
    
    predictions = model.predict(X_scaled)
    accuracy = accuracy_score(y, predictions)
    
    print("C =", C_value, "| Training accuracy =", accuracy)
```

Task 7.1
- Increasing the polynomial degree creates more polynomial features and allows the model to represent more complex relationships. However, it also increases model complexity and can increase the risk of overfitting.
- A very high polynomial degree can create a large number of features, making the model more computationally expensive and more complex. It can also cause overfitting, where the model performs well on the training data but poorly on unseen data.

PS: Polynomial features expand the original feature space by creating additional features based on polynomial combinations of the original features. In this experiment, the dataset increased from 2 features to 10 features. This allows the linear SVM to model more complex relationships in the original feature space.

```python
from sklearn.svm import SVC

poly_kernel_svm = SVC(
    kernel="poly",
    degree=3,
    coef0=1,
    C=5
)

poly_kernel_svm.fit(X_scaled, y)

y_poly_kernel_pred = poly_kernel_svm.predict(X_scaled)

poly_kernel_accuracy = accuracy_score(y, y_poly_kernel_pred)

print("Polynomial Kernel accuracy:", poly_kernel_accuracy)
```

Task 8.1
- The purpose of the polynomial kernel is to allow the SVM to model nonlinear relationships between the features and the classes. It does this without requiring us to explicitly create all the polynomial features ourselves.
- The degree parameter controls the degree of the polynomial used by the kernel.
- If the degree is too high, the model can become unnecessarily complex and may overfit the training data. It can also increase computational complexity.
- The kernel approach is more convenient because it allows the SVM to model polynomial relationships without explicitly creating and storing all the additional polynomial features.

```python
rbf_svm = SVC(
    kernel="rbf",
    C=1,
    gamma=1
)

rbf_svm.fit(X_scaled, y)

y_rbf_pred = rbf_svm.predict(X_scaled)

rbf_accuracy = accuracy_score(y, y_rbf_pred)

print("RBF Kernel accuracy:", rbf_accuracy)
```

```python
gamma_values = [0.01, 0.1, 1, 10, 100]

for gamma_value in gamma_values:

    model = SVC(
        kernel="rbf",
        C=1,
        gamma=gamma_value
    )

    model.fit(X_scaled, y)

    predictions = model.predict(X_scaled)

    print(
        "gamma =", gamma_value,
        "accuracy =", accuracy_score(y, predictions)
    )
```

Task 8.2
- When γ is very small, each observation has a broader influence on the decision boundary. The resulting boundary tends to be smoother and less complex. In the experiment, γ = 0.01 and γ = 0.1 both produced 95.33% training accuracy.
- When γ becomes large, the influence of individual observations becomes more localized. This allows the SVM to create a more complex and flexible decision boundary. In the experiment, γ = 10 produced 96.67% training accuracy, while γ = 100 produced 96.33%.
- A large γ makes the influence of each individual observation more localized. The model can therefore respond more strongly to individual observations and create a highly detailed or curved decision boundary.
- If γ is too large, the model can become excessively sensitive to individual training observations and fit the training data too closely. This can produce high training accuracy but poorer performance on unseen data, which is overfitting

```python
models = {
    "Linear SVM": svm_clf,
    "Polynomial Kernel": poly_kernel_svm,
    "RBF Kernel": rbf_svm
}
```

```python
from sklearn.model_selection import cross_val_score

for name, model in models.items():

    scores = cross_val_score(
        model,
        X_scaled,
        y,
        cv=5,
        scoring="accuracy"
    )

    print(
        name,
        "Mean accuracy:", scores.mean(),
        "Std:", scores.std()
    )
```

```python
lin_svr = make_pipeline(
    StandardScaler(),
    LinearSVR(
        epsilon=0.5,
        random_state=42,
        max_iter=10000
    )
)

lin_svr.fit(X_train, y_train)
```

```python
y_pred = lin_svr.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

Task 10.1
- SVM classification predicts a class or category, while SVM regression predicts a numerical value. In classification, the SVM tries to find a decision boundary that separates different classes. In regression, the objective is to predict a numerical value while controlling the size of the prediction errors.
- The ε (epsilon) parameter defines a tolerance region around the regression prediction. Errors that fall within this ε-tolerance are treated as being within the acceptable error range of the SVR.

```python
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

rbf_svr = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "svr",
        SVR(
            kernel="rbf",
            C=100,
            gamma="scale",
            epsilon=0.1
        )
    )
])
```

```python
rbf_svr.fit(X_train, y_train)
```

```python
y_pred = rbf_svr.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

print("RBF SVR RMSE:", rmse)
```

```python
C_values = [1, 10, 100, 1000]

for C_value in C_values:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("svr", SVR(
            kernel="rbf",
            C=C_value,
            gamma="scale",
            epsilon=0.1
        ))
    ])

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)

    print("C =", C_value, "| RMSE =", rmse)
```

```python
gamma_values = [0.001, 0.01, 0.1, 1, 10]

for gamma_value in gamma_values:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("svr", SVR(
            kernel="rbf",
            C=100,
            gamma=gamma_value,
            epsilon=0.1
        ))
    ])

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)

    print("gamma =", gamma_value, "| RMSE =", rmse)
```

```python
epsilon_values = [0.01, 0.1, 0.5, 1.0]

for epsilon_value in epsilon_values:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("svr", SVR(
            kernel="rbf",
            C=100,
            gamma="scale",
            epsilon=epsilon_value
        ))
    ])

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)

    print(
        "epsilon =", epsilon_value,
        "| RMSE =", rmse
    )
```
